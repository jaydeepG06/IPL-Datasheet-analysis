```python
df20 = df[(df['batter'] == 'V Kohli') & (df['striker_out'] == True)].groupby('bowling_team').agg({'batter_balls': 'sum',
                                                                                                 'batter_runs': 'sum'})
```

```python
df20['strike_rate'] = ((df20['batter_runs'] / df20['batter_balls'])* 100)
```

```python
df20 = df20.groupby('bowling_team').agg({'strike_rate': 'mean'})
```

```python
df20
```

```python
import matplotlib.pyplot as plt
df20['strike_rate'].sort_values().plot(kind='barh',
    figsize=(10,6)
)
plt.xlabel('Strike_rate')
plt.ylabel('bowling_team')
plt.title('Virat Kohli Strike Rate vs Opposition')
plt.show()
```

```python
plt.plot(
    df20.index,
    df20['strike_rate'],
    marker='o'
)

for x, y in zip(df20.index, df20['strike_rate']):
    plt.text(x, y, round(y, 2))

plt.xticks(rotation=90)
plt.show()
```

```python
df.reset_index()
ROKO = df[df['batter'].isin(['V Kohli', 'R Sharma','RG Sharma'])]
ROKO
```

```python
ROKO = ROKO.groupby(['bowling_team', 'batter']).agg({'batter_balls': 'sum',
                                          'batter_runs': 'sum'})
```

```python
ROKO['strike_rate'] = (ROKO['batter_runs'] / ROKO['batter_balls'])*100
```

```python
ROKO1 = ROKO.groupby(['bowling_team','batter']).agg({'strike_rate': 'mean'})
```

```python
ROKO1 = ROKO1.reset_index()
ROKO1
```

```python
ROKO1['batter'] = ROKO1['batter'].replace('R Sharma ', 'RG Sharma')
```

```python
kohli = ROKO1[ROKO1['batter'] == 'V Kohli']
dhoni = ROKO1[ROKO1['batter'] == 'RG Sharma']

plt.figure(figsize=(10,6))

plt.plot(kohli['bowling_team'],
         kohli['strike_rate'],
         marker = 'o',
         label = 'Kohli'
)

plt.plot(dhoni['bowling_team'],
         dhoni['strike_rate'],
         marker = 'o',
         label = 'Rohit'
)
for x, y in zip(kohli['bowling_team'], kohli['strike_rate']):
    plt.text(x, y, round(y, 2), fontsize=8)

# Labels for Rohit
for x, y in zip(dhoni['bowling_team'], dhoni['strike_rate']):
    plt.text(x, y, round(y, 2), fontsize=8)
    
plt.xlabel("Opponent")
plt.ylabel("Strike Rate")
plt.title("Strike Rate vs Opposition")
plt.legend()
plt.grid(True)
plt.xticks(rotation = 90)

plt.show()
```
